Find S algorithm :
Algorithm that finds most specific hypothesis consistent with positive examples.

```python
def find_s_also(training_data):
    hypo=None
    for instance in training_data:
        features=instance[:-1]
        label=instance[-1]
        if label=='Yes':
            if hypo == None:
                hypo=list(features)
            else:
                for i in range(len(hypo)):
                    if hypo[i]!=features[i]:
                        hypo[i]='?'
    return hypo

training_data=[['Sunny','Warm','Normal','Strong','Warm','Same','Yes'],
               ['Sunny','Warm','High','Strong','Warm','Same','Yes'],      
               ['Rainy','Cold','High','Strong','Warm','Change','No'],      
               ['Sunny','Warm','High','Strong','Cool','Change','Yes']]

print(find_s_also(training_data))
```
