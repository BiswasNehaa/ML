```python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
import matplotlib.pyplot as plt

# Generate 100 random values between 0 and 1
np.random.seed(42)
X = np.random.rand(100, 1)

# Label first 50 points
# if x <= 0.5 → Class 0, else → Class 1
y = np.zeros(100, dtype=int)
for i in range(50):
    if X[i] <= 0.5:
        y[i] = 0  # Class 1
    else:
        y[i] = 1  # Class 2

# Training data = first 50 points
X_train = X[:50]
y_train = y[:50]

# Test data = remaining 50 points
X_test = X[50:]
y_test = y[50:]

# Try different values of k
k_values = [1, 2, 3, 4, 5, 20, 30]

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    y_pred = knn.predict(X_test)
    accuracy = np.mean(y_pred == y_test) * 100
    print(f"k={k}: Accuracy = {accuracy:.2f}%")

# Visualize for k=3
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)
y_pred = knn.predict(X_test)

plt.figure(figsize=(10, 4))
plt.scatter(X_train, y_train, color='blue', label='Training data', marker='o')
plt.scatter(X_test, y_pred, color='red', label='Predicted (k=3)', marker='x')
plt.xlabel("X values")
plt.ylabel("Class")
plt.title("KNN Classification (k=3)")
plt.legend()
plt.show()
```
