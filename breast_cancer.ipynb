```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

data=load_breast_cancer()
x=data.data
y= data.target

#split
x_train,x_test,y_train, y_test = train_test_split (x, y, test_size=0.2, random_state=42)

#train
dt=DecisionTreeClassifier(max_depth=3, random_state=42)
dt.fit(x_train, y_train)

#predict
y_pred= dt.predict(x_test)
print(f"Accyracy test: {accuracy_score(y_test,y_pred):.4f}")

new_sample= x_test[0].reshape(1,-1)
prediction=dt.predict(new_sample)
print (f" Predicted class :{'melignant' if prediction[0]==0 else 'benign' }")

plt.figure(figsize=(20,8))
plot_tree(dt,filled=True, feature_names=data.feature_names, class_names=data.target_names, fontsize=8 )
plt.title("precufesfj")
plt.show()
```
